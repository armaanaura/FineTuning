```python
with open('../tinyshakespeare.txt', 'r', encoding = 'utf-8') as f:
    input_text = f.read()
```

Computers store everything as **bits (0s and 1s)**.

- Numbers can be stored directly in binary.
- Text needs an **encoding** to convert characters into numbers/bytes.
- Example: character `'1'` in UTF-8 is stored as `00110001`, not `00000001`.
- Common text encodings are **ASCII, UTF-8, UTF-16, and UTF-32**.
- **UTF-8** is the most widely used today because it supports almost all languages and symbols.

```python
character_set = set(input_text)
print(len(character_set))
```

```python
characters = sorted(list(character_set))
vocab_size = len(character_set)
print("size of the character set is: " + str(len(characters)))
print(''.join(characters))
```

```python
print(vocab_size)
```

stoi (String-to-Integer): A dictionary or function that maps a text character or word to a unique number ID.Example: {'a': 1, 'b': 2, 'c': 3} meaning stoi['b'] returns 2.itos (Integer-to-String): The reverse dictionary or function that maps the number ID back to the original text character or word.Example: {1: 'a', 2: 'b', 3: 'c'} meaning itos[2] returns 'b'.

```python
stoi = {}
itos = {}

for i,ch in enumerate(characters):
    stoi[ch] = i
    itos[i] = ch
```

```python
print(stoi)
print(itos)
```

```python
encode = lambda string : [stoi[ch] for ch in string]
decode = lambda listOfIntegers : "".join([itos[num] for num in listOfIntegers])

print(encode("I am armaan"))
temp = [21, 1, 39, 51, 1, 39, 56, 51, 39, 39, 52]
print(decode(temp))
```

```python
encoded_input = encode(input_text)
```

```python
print(encoded_input[:100])
```
